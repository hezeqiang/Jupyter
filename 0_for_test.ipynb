```python
def fact(n):
    '''
    Calculate 1*2*...*n
    
    >>> fact(1)
    1
    >>> fact(10)
    
    >>> fact(-1)
    Traceback (most recent call last):
        ...
    KeyError: 'empty'
    '''
    if n < 1:
        raise ValueError()
    if n == 1:
        return 1
    return n * fact(n - 1)
```
