```python
# sqrt(x^2+y^2)

points = np.arange(-5, 5, 0.01) # 1000 equally spaced points
xs, _ = np.meshgrid(points, points)

points = np.arange(5, 15, 0.01) # 1000 equally spaced points
_, ys = np.meshgrid(points, points)


print(xs, "\n",ys)
z = np.sqrt(xs ** 2 + ys ** 2)
z
import matplotlib.pyplot as plt
plt.imshow(z, cmap=plt.cm.gray); plt.colorbar()
plt.title("Image plot of $\sqrt{x^2 + y^2}$ for a grid of values")
```

```python
import random
position = 0
walk = [position]
steps = 1000
for i in range(steps):
     step = 1 if random.randint(0, 1) else -1
     position += step
     walk.append(position)
        
plt.plot(walk[:100])  
```

```python
nsteps = 1000
draws = np.random.randint(0, 2, size=nsteps)
steps = np.where(draws > 0, 1, -1)
walk = steps.cumsum()
plt.plot(walk[:100])   
```

```python
nwalks = 4
nsteps = 200
draws = np.random.randint(0, 2, size=(nwalks, nsteps)) # 0 or 1
print(draws)
steps = np.where(draws > 0, 1, -1)

walks = steps.cumsum(1)
walks
```

```python
hits10 = (np.abs(walks) >= 10).any(1)
print(hits10)
hits10.sum() # Number that hit 30 or -30
```

```python
print(np.abs(walks[hits10]))
print((np.abs(walks[hits10]) >= 10).argmax(1))
crossing_times = (np.abs(walks[hits10]) >= 10).argmax(1)
print(crossing_times)
```
