# Latex -> pdf

## 使用python

直接在winEdt中点击最下方的Windows Command Prompt, 否则从cmd进入命令行的话，还要进入tex文件夹，好麻烦。

1、tex ——> dvi

> latex XX.tex

其中，XX为编辑文件名

2、dvi ——> pdf

> dvipdfm XX.dvi

open('tese.tex','w',encoding='utf-8')
对于复杂的文件，可以保存如
file_suffix='.tex'
open('tese'+'new'+'naming'+'approach'+file_suffix,'w',encoding='utf-8')

## 使用bat文件


latex --synctex=1 filename(like note)
::bibtex OneReport >nul     (无效的注释，命令行不输出任何文字)
latex --synctex=1 OneReport >nul  (不输出任何文字)
::latex --synctex=1 OneReport >nul   (无效的注释，命令行不输出任何文字)
dvipdfm filename(like note)       

>nul 表示shell中不输出任何文字
:: 表示注释

## 实例

```python
import os
file_suffix='.tex'
fname=open('tese.tex','w',encoding='utf-8')
fname_complex=open('tese'+'new'+'naming'+'approach'+file_suffix,'w',encoding='utf-8')
fname_complex.close()
print ('Hello, World！',file=fname)
fname.close()
os.system("latex note.tex")
os.system("dvipdfm note")
```
