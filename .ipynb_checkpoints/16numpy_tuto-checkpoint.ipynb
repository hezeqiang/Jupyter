```python
# sqrt(x^2+y^2)

points = np.arange(-5, 5, 0.01) # 1000 equally spaced points
xs, _ = np.meshgrid(points, points)

points = np.arange(5, 15, 0.01) # 1000 equally spaced points
_, ys = np.meshgrid(points, points)


print(xs, "\n",ys)
z = np.sqrt(xs ** 2 + ys ** 2)
z
import matplotlib.pyplot as plt
plt.imshow(z, cmap=plt.cm.gray); plt.colorbar()
plt.title("Image plot of $\sqrt{x^2 + y^2}$ for a grid of values")
```

```python
arr = np.random.randn(5, 4)
arr.mean()
np.mean(arr)
arr.sum()
print(
arr.mean(axis=1), "\n",
arr.sum(axis=0))
```

```python
arr = np.arange(0,12).reshape(3,4)
print(arr.cumsum(axis=0)) # accumulative sum

print(arr.mean())
print(arr.min())
print(arr.std())
print(arr.var())
print(arr.argmax())
```
