# 调用 ，产生 打印 保存文件

```python
import json

# bulid a dictionary of json 
sampleJson1 = {"id" : 1, "name" : "value2", "age" : 29}

# print the json dictionary in pretty way
prettyPrintedJson  = json.dumps(sampleJson1, indent=2, separators=(",", " = "))
print(prettyPrintedJson)

# save the jspn dictionary into a json file
print("Started writing JSON data into a file")
with open("sampleJson1.json", "w") as write_file:
    json.dump(sampleJson1, write_file, indent=4, sort_keys=True)
print("Done writing JSON data into a file")
```

# 访问嵌套的value

```python
import json

sampleJson2 = """{
   "company":{
      "employee":{
         "name":"emma",
         "payble":{
            "salary":7000,
            "bonus":800
         }
      }
   }
}"""

data = json.loads(sampleJson2)
print(data['company']['employee']['payble']['salary'])

# save the jspn dictionary into a json file
print("Started writing JSON data into a file")
with open("sampleJson1.json", "w") as write_file:
    json.dump(sampleJson2, write_file, indent=4, sort_keys=True)
print("Done writing JSON data into a file")
```

# 类对象和 json 转换

```python
# 保存 对象 到 json 文件
import json
from json import JSONEncoder

class Vehicle:
    def __init__(self, name, engine, price):
        self.name = name
        self.engine = engine
        self.price = price

class VehicleEncoder(JSONEncoder):
        def default(self, o):
            return o.__dict__

vehicle = Vehicle("Toyota Rav4", "2.5L", 32000)

print("Encode Vehicle Object into JSON")
vehicleJson = json.dumps(vehicle, indent=4, cls=VehicleEncoder)
print(vehicleJson)



# 读取 json 文件产生对象

import json

class Vehicle:
    def __init__(self, name, engine, price):
        self.name = name
        self.engine = engine
        self.price = price

def vehicleDecoder(obj):
        return Vehicle(obj['name'], obj['engine'], obj['price'])

vehicleObj = json.loads('{ "name": "Toyota Rav4", "engine": "2.5L", "price": 32000 }',
           object_hook=vehicleDecoder)

print("Type of decoded object from JSON Data")
print(type(vehicleObj))
print("Vehicle Details")
print(vehicleObj.name, vehicleObj.engine, vehicleObj.price)
```
