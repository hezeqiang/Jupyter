# Magnetic field calculation

## 1 axis displacement

```python
from IPython.display import display
import control as ctl
import os
import matplotlib as mpl
from matplotlib.pyplot import subplots
import matplotlib.pyplot as plt
import numpy as np
from sympy import *

# only in x axis diaplacement

theta, x, MMF= symbols('theta x F_m ')

Br, F, h, R = symbols( ' B_theta F_x  h  R ')

MMF=4.75*cos(21*theta)

Br=MMF/(9.5-cos(theta)*x)

F=Integral(Br**2*R*h*pi/2/4/pi/(10**-7), (theta, 0,2*pi))

# expr.subs([(x, 2), (y, 4), (z, 0)])
display(F)

display(F.subs(x, 1))

# display(F.doit())
```

## 2 axes displacement

```python
from IPython.display import display
import control as ctl
import os
import matplotlib as mpl
from matplotlib.pyplot import subplots
import matplotlib.pyplot as plt
import numpy as np
from sympy import *

# only in x axis diaplacement

theta, x,y, MMF= symbols('theta x y F_m ')

Br, Fx, Fy,h, R = symbols( ' B_theta F_x  F_y h  R ')

lg, B, lpm ,u= symbols( ' l_g \hat{B} l_{pm} u_0')

MMF=(lg+lpm)*B*cos(21*theta)

Br=MMF/((lg+lpm)-cos(theta)*x)

Fx=Integral(Br**2*R*h*pi/2/u*cos(theta), (theta, 0,2*pi))

# expr.subs([(x, 2), (y, 4), (z, 0)])
display(Fx)

display(Fx.subs(x, 1))

# display(F.doit())
print(latex(Fx))

Br=MMF/((lg+lpm)-cos(theta)*y)

Fy=Integral(Br**2*R*h*pi/2/u*sin(theta), (theta, 0,2*pi))
display(Fy)

display(Fy.subs(y, 1))

# display(F.doit())
print(latex(Fy))
```
