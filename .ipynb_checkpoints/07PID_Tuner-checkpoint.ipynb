# PID Tuner

```python
from IPython.display import display
from sympy import *
import control as ctl
init_printing(use_latex='mathjax')

def make_poly(N,coeffs):
    """Create a polynomial in x."""
    coeffs = list(reversed(coeffs))
    y = 0
    for i in range(N):
        y += (x**i)*coeffs[i]
    y = sym.poly(y)
    return y

# Symbols
Kp, Ki , p, R, L= symbols('K_p K_i p R L')

PI = Kp  + Ki / p
motor_winding = 1/(R+p*L)
G_open = PI * motor_winding
G_closed = G_open / (1 + G_open)

display(G_closed)
G_closed=simplify(G_closed)
print(type(G_closed))
num, den = fraction(G_closed)

num=expand(num)
den=expand(den)

Poly_num=display(Poly(num, p).coeffs())
Poly_den=display(Poly(den, p).coeffs())

print(type(Poly(num, p).coeffs()))

display('Numerator', num)
display('Denominator', den)

display((num/den)) # expand() and factor() ceiling()

print(latex(Poly(num,p)))
print(latex(Poly(den,p)))
```

```python
import os
import matplotlib as mpl
from matplotlib.pyplot import subplots
import matplotlib.pyplot as plt
import numpy as np

def get_coeffs_motor_current_regulator(R, L, Bandwidth_Hz):
    
    Ki = 2 * np.pi * Bandwidth_Hz *R
    Kp = Ki * L / R
    return Kp, Ki

import control
from control import bode_plot

# R = 3 # Ohm
R = 0.2 # Ohm
L = 0.004 # H

Kp, Ki = get_coeffs_motor_current_regulator(R, L, 1200)
# force the closed loop tf to have a pair of complex pole and observe its step response
print('Kp =', Kp, '|', 'Ki =', Ki)
display('Kp is', Kp)
display('Ki is', Ki)

motor = control.tf([1], [L, R])
pi_regulator = control.tf([Kp, Ki], [1, 0])
display(motor)
display(pi_regulator)
# figure()
# mag, phase, omega = bode_plot(dc_motor, 2*np.pi*np.logspace(-2,4,1000), dB=1, Hz=1, deg=1)
# figure()
# mag, phase, omega = bode_plot(pi_regulator, 2*np.pi*np.logspace(-2,4,1000), dB=1, Hz=1, deg=1)

open_sys = control.series(pi_regulator, motor)
closed_sys = control.feedback(open_sys, 1, sign=-1)

# open_sys = pi_regulator * dc_motor
# closed_sys = open_sys / (1+open_sys)

display(open_sys)
display(closed_sys)
closed_sys = control.minreal(closed_sys) # cancel out the poles and zeros
display(closed_sys)
print('Ki/R = w0 = 2*pi* Bandwidth_Hz=', Ki/R)
print('Ki/Kp = R/L', R/L)

# figure()
# mag, phase, omega = bode_plot(open_sys, 2*np.pi*np.logspace(-2,4,1000), dB=1, Hz=1, deg=1)
plt.figure()
mag, phase, omega = bode_plot(closed_sys, 2*np.pi*np.logspace(-2,4,1000), dB=1, Hz=1, deg=1)

# T, yout = control.step_response(closed_sys, np.arange(0,0.02,1e-5))
T, yout = control.step_response(closed_sys, np.arange(0,0.02,1e-5))
plt.figure()
plt.plot(T, yout)
plt.xlabel('time [s]')
plt.xlim([0,0.005])
plt.ylim([0,1.2])
plt.legend(loc='upper right')

# 上升时间： 上升到0.9的程度
print(
    min(yout, key=lambda x:abs(x-0.9)), 'A'
) 
print(
    T[(np.abs(yout-0.9)).argmin()] * 1000, 'ms' 
)

print('dc gain:', control.dcgain(closed_sys))
print('zero:', control.zero(closed_sys))
print('pole:', control.pole(closed_sys))

```

# For Fun: Varying Parameters

```python
# R = 3 # Ohm
R = 0.2 # Ohm
L = 0.004 # H

list_zeros = []
list_poles = []
for L in np.arange(0.001, 0.009, 0.002):  # numpy.arange(start, stop, step, dtype = None)
    Kp, Ki = get_coeffs_motor_current_regulator(R, L, 200)
    print(f'Kp = {Kp:g} | Ki = {Ki:g}')
    motor = control.tf([1], [L, R])
    pi_regulator = control.tf([Kp, Kp*Ki], [1, 0])
    open_sys = control.series(pi_regulator, motor)
    closed_sys = control.feedback(open_sys, 1, sign=-1)

    closed_sys = control.minreal(closed_sys)
    
    list_zeros.append( control.zero(closed_sys) )
    list_poles.append( control.pole(closed_sys) )

for poles, zeros in zip(list_poles, list_zeros):
    print('poles:', poles, '| zeros:', zeros)
    plt.plot(poles.real, poles.imag, 'o')
```

# Speed PI Regulator

```python
from IPython.display import display
from sympy import *
init_printing(use_latex='mathjax')

# From current loop
p, L, Tc = symbols('p L T_c')
Gi_closed = 1 / (1+Tc*p)

# Symbols
speedKp, speedKi, J_s = symbols('K^\omega_p K^\omega_i J_s')

speedPI = speedKp  + speedKi / p
motor_motion = 1/ J_s /p
G_open = motor_motion * Gi_closed * speedPI
G_open=simplify(G_open)
display('Open tf:', G_open)
print(latex(G_open))
# display(simplify(G_open))

G_closed = G_open / (1 + G_open)

display('Closed tf:', G_closed)
# print(latex(G_closed))
display(simplify(G_closed))
print(latex(simplify(G_closed)))
# print(type(G_closed))

num, den = fraction(G_closed)
display('Numerator', num)
display('Denominator', simplify(den))

# display(Poly(num, p).coeffs())
# display(Poly(simplify(den), p).coeffs())

# print(latex(Poly(num,p)))
# print(latex(Poly(simplify(den),p)))


print('\n','the determination of controller:')

G_sub=speedKp*(speedKp/speedKi*p+1)*Gi_closed
display(G_sub)

T1, T2, eta, omegac = symbols('T_1,T_2 \eta \omega_c')

T2=Tc

speedKp=J_s/eta/(T2**2)*sqrt((1/eta+1)/(eta+1))
speedKi=J_s*sqrt((1/eta+1)/(eta+1))/T2

display(speedKp)
display(speedKi)
print(latex(speedKp))
print(latex(speedKi))
```

```python
import matplotlib.pyplot as plt
import control as ctl

R = 0.5 # Ohm
L = 0.003 # H
Kp, Ki = get_coeffs_motor_current_regulator(R, L, 1200)
Tc=R/Ki
Gi_closed = control.tf([1], [Tc, 1]) 
currentBandwidth_radPerSec =1/ Tc
print('BW_i:', currentBandwidth_radPerSec/(2*np.pi), 'Hz')

def get_coeffs_motor_SPEED_regulator(J_s, eta, Tc):
    speedKi = J_s/eta/(Tc**2)*sqrt((1/eta+1)/(eta+1))
    speedKp = J_s*sqrt((1/eta+1)/(eta+1))/Tc
    return speedKp, speedKi

J_s  = 0.06
motor_motion = control.tf([1/J_s], [1, 0])
display('motion:', motor_motion)

eta = 3 # damping factor (nothing similar to damping ratio for 2nd-order tf)

fig1, fig2, fig3 = plt.figure(1), plt.figure(2), plt.figure(3)

# if True:
for eta in [1.5, 2, 4, 6, 8, 10, 20]:

    speedKp, speedKi = get_coeffs_motor_SPEED_regulator(J_s, eta, Tc)
#   speedKp *= 1000
    print(f'speedKp = {speedKp:g}', f'speedKi = {speedKi:g}',f'Kp = {Kp:g}', f'Ki = {Ki:g}', f'cutoff = {1/sqrt(eta)/Tc/2/np.pi} Hz', sep=' | ')
    print([speedKp, speedKi])
    speedPI = ctl.tf([int(speedKp), int(speedKi)], [1, 0])

    Gw_open = motor_motion * Gi_closed * speedPI
#     display('speed open loop:', Gw_open)
#     print(control.dcgain(Gw_open), 
#           control.zero(Gw_open)/2/np.pi,
#           control.pole(Gw_open)/2/np.pi, sep='\n')

    plt.figure(1)
    mag, phase, omega = bode_plot(Gw_open, 2*np.pi*np.logspace(0,4,500), dB=1, Hz=1, deg=1, lw='0.5', label=f'{eta:g}')
#     gm,pm,sm,wg,wp,ws = control.stability_margins(Gw_open)
#     print(f'GM={gm}, PM={pm:g} deg',wg,'Hz',wp,'Hz')

    Gw_closed = Gw_open / (1+Gw_open)

    plt.figure(2)
    mag, phase, omega = bode_plot(Gw_closed, 2*np.pi*np.logspace(0,4,500), dB=1, Hz=1, deg=1, lw='0.5', label=f'{eta:g}')

    plt.figure(3)
    T, yout = control.step_response(Gw_closed, np.arange(0,0.05,1e-5))
    # 上升时间：
    print('\t', 
        min(yout, key=lambda x:abs(x-0.9)), 'A'
    )
    rise_time_ms = T[(np.abs(yout-0.9)).argmin()] * 1000
    print('\t', 
        rise_time_ms, 'ms' 
    )
    plt.plot(T, yout, lw='0.5', label=f'{eta:g} | {rise_time_ms:g} ms')


fig2.axes[0].set_ylim([-3, 20]) # -3dB
fig3.axes[0].set_xlabel('time [s]')
fig3.axes[0].set_ylabel('speed [elec. deg]')

for i in [1,2,3]:
    plt.figure(i)
    plt.legend(loc='upper right')
```

```python
# Iterate for a preset speed rise time and delta

# Pick your favorite step repsonse by "shape" via damping factor eta
eta = 4

# Specify your desired speed step response rise time
desired_rise_time_ms = 2.3

rise_time_ms = 1000 # ms (initial)
BW_current_Hz = 500 # Hz (initial)
while True:
    # Current loop (Tune its bandwidth to support required speed response)
    if abs(rise_time_ms - desired_rise_time_ms)<=1: # ms
        break
    else:
        if rise_time_ms > desired_rise_time_ms:
            BW_current_Hz += 10 # Hz
        else:
            BW_current_Hz -= 10 # Hz
            if BW_current_Hz<=0:
                raise Exception('Change the step size of "BW_current_Hz" and try again.')
                break
    print(f'BW_current_Hz = {BW_current_Hz}')
    R = 0.5 # Ohm
    L = 0.017 # H
    Kp, Ki = get_coeffs_motor_current_regulator(R, L, BW_current_Hz)
    Tc=R/Ki
    Gi_closed = control.tf([1], [Tc, 1]) 
    currentBandwidth_radPerSec = Kp/L

    # Speed loop
    J_s  = 0.06
    motor_motion = control.tf([1/J_s], [1, 0])
    speedKp, speedKi = get_coeffs_motor_SPEED_regulator(J_s, eta, Tc)
#   speedKp *= 1000
    print(f'speedKp = {speedKp:g}', f'speedKi = {speedKi:g}',f'Kp = {Kp:g}', f'Ki = {Ki:g}', f'cutoff = {1/sqrt(eta)/Tc/2/np.pi} Hz', sep=' | ')
 
    speedPI = ctl.tf([int(speedKp), int(speedKi)], [1, 0])
    Gw_open = motor_motion * Gi_closed * speedPI
    Gw_closed = Gw_open / (1+Gw_open)

    T, yout = control.step_response(Gw_closed, np.arange(0,0.05,1e-5))
    rise_time_ms = T[(np.abs(yout-0.9)).argmin()] * 1000 # <- Potential bug here! rise time is not always correct.

plt.figure()
plt.plot(T, yout, lw='0.5', label=f'{eta:g} | {rise_time_ms:g} ms')
plt.xlabel('time [s]')
plt.ylabel('speed [elec. deg / sec]')
plt.legend(loc='upper right')
```
